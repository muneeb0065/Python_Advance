### **Basically two types of files:**
**Text Files:** All program files are text files <br>
**Binary Files:** Images, videos or .exe files......

How File I/O is done in most programming languages:
1. Open File
2. Read or Write Data
3. Close the File

```python
# Casw 1: Create a file is not present

file = open('sample.txt', 'w')
file.write("I write something in the file")
file.close()

file.write("some more changes.") #since we have closed the file so this will not work
```

```python
# write multiline strings
f = open('sample1.txt','w')
f.write('hello world')
f.write('\nhow are you?')
f.close()
```

```python
# Writing in an existing file

f = open('sample1.txt' , 'w')
f.write("New content added and previously will be erased")

f.close()
```

So you can see the 'w' mode deletes the previous writing and overwrites so to tackle this we hav append mode 'a'.

```python
# opening a file in append mode

f = open('sample.txt', 'a')
f.write('\nI am writing in this additional line')
f.close()
```
